```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import scipy 
from scipy import stats
```

```python
mean_p=98.6
mean_s=mean_p
size1=10000
n     =36
std_p =0.73
std_s =std_p/np.sqrt(n)
x_s   =98.3
zscore_s=(x_s-mean_s)/std_s
prob    = stats.norm.cdf(zscore_s)*100
print("{:.2f}%".format(prob))
np.random.seed(2)
xp=(np.random.normal(loc=mean_p,scale=std_p,size=size1))
xs=(np.random.normal(loc=mean_s,scale=std_s,size=size1))

sns.kdeplot(xp,label="Population Data",ls="--")
sns.kdeplot(xs,label="Sample Means",ls="--")
plt.legend(facecolor="cyan")
plt.annotate("std_s: {:.2f}\nstd_p: {:.2f}\nmean_p: {}\nmean_s :{}".format(std_s,std_p,mean_p,mean_s),xy=(95.5,2.4),size=15,color="blue")
plt.annotate("$\\sigma_{\\bar x}=\\frac{\\sigma}{\\sqrt{n}}$",xy=(95.5,1),color="red",size=20)
plt.annotate("$\\mu_{\\bar x}=\\mu$",xy=(95.5,0.5),color="green",size=20)
plt.title("Central Limit Theorem",x=0.28,size=30,color="indigo")
plt.xlabel("Body Temperature of Adults in Degrees $\\degree $ F",color="brown")
plt.ylabel("Distribution Probability")
plt.axis("off")
plt.show()
```

```python
mean_p=98.6
mean_s=mean_p
size1=10000
n     =36
std_p =0.73
std_s =std_p/np.sqrt(n)
x_s   =98.3
zscore_s=(x_s-mean_s)/std_s
prob    = stats.norm.cdf(zscore_s)*100
print("{:.2f}%".format(prob))
np.random.seed(2)
xp=np.random.normal(loc=mean_p,scale=std_p,size=size1)
xs=np.random.normal(loc=mean_s,scale=std_s,size=size1)

sns.kdeplot(xp,label="Population Data",cumulative=True,ls="--")
sns.kdeplot(xs,label="Sample Means",cumulative=True,ls="--")
plt.legend(facecolor="cyan")
plt.annotate("std_s: {:.2f}\nstd_p: {:.2f}\nmean_p: {}\nmean_s :{}".format(std_s,std_p,mean_p,mean_s),xy=(95.5,2.4),size=15,color="blue")
plt.annotate("$\\sigma_{\\bar x}=\\frac{\\sigma}{\\sqrt{n}}$",xy=(95.5,0.6),color="red",size=20)
plt.annotate("$\\mu_{\\bar x}=\\mu$",xy=(95.5,0.5),color="green",size=20)
plt.title("Central Limit Theorem",x=0.28,size=30,color="indigo")
plt.xlabel("Body Temperature of Adults in Degrees $\\degree $ F",color="brown")
plt.ylabel("Distribution Probability")
# plt.axis("off")
plt.show()
```

```python
mean_p=98.6
size1=1000000
n     =360
freq_s=500
std_p =0.73
np.random.seed(2)
xp=(np.random.normal(loc=mean_p,scale=std_p,size=size1))
fig,ax=plt.subplots(1,1,figsize=(12,6))
means=[]
for i in range(freq_s):
    means.append(np.mean(np.random.choice(xp,n,replace=True)) )
sns.kdeplot(xp,label="Population Data Distribution")
sns.kdeplot(means,label="Sample Means Distribution")
plt.legend(facecolor="beige",loc=(0,1.02),ncol=2)
plt.annotate("sample size {} ,\nsample frequency ={} \n{} means".format(n,freq_s,freq_s),
             size=15,
             xytext=(95.5,ax.get_ylim()[1]*0.8),
             xy=(mean_p-0.1,ax.get_ylim()[1]*0.07),
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="cyan",edgecolor="red"),
             arrowprops=dict(arrowstyle="fancy",connectionstyle="arc3,rad=0.3")
            )
plt.title("Central Limit Theorem-Temperature Quiz",x=0.45,y=1.1,size=16,color="indigo")
plt.xlabel("Body Temperature of Adults in Degrees $\\degree $ F",color="brown")
plt.ylabel("Distribution Probability")
plt.savefig('sd.pdf')
plt.show()
```
