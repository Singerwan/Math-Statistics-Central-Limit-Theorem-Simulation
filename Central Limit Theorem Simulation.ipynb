```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(2)
population= np.random.randint(low=100,high=120,size=10000000)

sample_size= 200   # sample size n >30
sample_time= 300 
lst=[]
for i in range(1,sample_time,1):
    lst.append(np.mean([np.random.choice(population,sample_size,replace=True)]))
sns.kdeplot(lst)
print(lst[:2])
plt.show()
```
